- R_DEBT_INCOME: Rasio antara hutang terhadap pendapatan.

- R_DEBT_SAVINGS: Rasio antara hutang terhadap tabungan.

- R_CLOTHING, R_CLOTHING_INCOME: Rasio pengeluaran untuk pakaian terhadap tabungan dan pendapatan.

- R_CLOTHING_SAVINGS, R_CLOTHING_DEBT: Rasio pengeluaran pakaian terhadap tabungan dan hutang.

- T_EDUCATION_12, T_EDUCATION_6: Pengeluaran nasabah untuk pendidikan dalam 12 bulan dan 6 bulan terakhir.

- R_EDUCATION, R_EDUCATION_INCOME: Rasio pengeluaran pendidikan terhadap pendapatan dan tabungan.

- T_ENTERTAINMENT_12, T_ENTERTAINMENT_6: Pengeluaran nasabah untuk hiburan dalam 12 bulan dan 6 bulan terakhir.

- R_ENTERTAINMENT, R_ENTERTAINMENT_INCOME: Rasio pengeluaran hiburan terhadap pendapatan dan tabungan.

- T_FINES_12, T_FINES_6: Pengeluaran nasabah untuk denda dalam 12 bulan dan 6 bulan terakhir.

- R_FINES, R_FINES_INCOME: Rasio pengeluaran denda terhadap pendapatan dan tabungan.

```python
# Drop kolom ID karena tidak memiliki nilai prediktif
df = df.drop(columns=["CUST_ID"])
```

```python
# Menentukan target dan fitur
target_col = "DEFAULT"

X = df.drop(columns=[target_col])
y = df[target_col]
```

```python
# Encoding fitur kategorikal
X_encoded = pd.get_dummies(X, drop_first=True)
```

```python
from sklearn.preprocessing import StandardScaler

scaler = StandardScaler()
X_scaled = scaler.fit_transform(X_encoded)
```

```python
processed_df = pd.DataFrame(
    X_scaled,
    columns=X_encoded.columns
)

processed_df[target_col] = y.values
processed_df.head()
```

```python
processed_df.to_csv("credit_score_preprocessed.csv", index=False)
```
